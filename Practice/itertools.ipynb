```python
from itertools import takewhile
def is_positive(num):
    return num > 0
x = [3, 4, 10, -2, 200]

result = list(takewhile(is_positive, x))
print(result)
```

```python
from itertools import takewhile
def is_positive(num):
    return num < 0
x = [3, 4, 10, -2, 200, -1, 0]

result = list(takewhile(is_positive, x))
print(result)
```
