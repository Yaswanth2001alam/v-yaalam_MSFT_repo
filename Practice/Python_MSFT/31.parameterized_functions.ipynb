force_uppercase is an parameter passed to an function

```python
def get_initial1(name, force_uppercase = True):
    if force_uppercase:
        initial = name[0:3].upper()
    else:
        initial = name [0:5]
    return initial

first_name2 = input("Enter the first name: ")
print(first_name2)
first_name_test = get_initial1(first_name2)

print('Your initial is:  '+ first_name_test)
```

```python
def get_initial2(name, first_uppercase):
    if first_uppercase:
        initial5 = name[0:3].upper()
    else:
        initial5 = name[0:6]
    return initial5

first_name6 = input("enter the firstname6: ")
print(first_name6)
first_name_init = get_initial2(first_uppercase=True, name=first_name6)
print("Your initial is: " + first_name_init)
```
