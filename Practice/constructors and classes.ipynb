```python
class Speed:
    def __init__(self, speed, auto, gear):
        self.speed = speed
        self.auto = auto 
        self.gear = gear

yas = Speed(70, 80, "yaswanth")
alam = Speed("network", "automation", "Engineer")

print(f"The speed was {yas.gear}")
print(f"The name is alam and the name was {alam.auto}")
```

```python
class Speed:
    def __init__(self, speed, auto, gear):
        self.speed = speed
        self.auto = auto 
        self.gear = gear
    def double_speed(self):
        self.speed *= 2
        

yas = Speed(70, 80, "yaswanth")
alam = Speed("network", "automation", "Engineer")

print(f"The speed was : {yas.gear}")
print(f"The name is alam and the name was : {alam.auto}")

yas.double_speed()

print(f"The value was :  {yas.speed}")
```
